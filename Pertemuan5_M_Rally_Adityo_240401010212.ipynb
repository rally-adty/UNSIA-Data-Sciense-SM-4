```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme(
    style='whitegrid',
    palette='Set2'
)
```

```python
df = sns.load_dataset('iris')
```

```python
x = [1,2,3,4,5]

y = [4,7,3,9,5]

fig, ax = plt.subplots(figsize=(8,5))

ax.plot(
    x,
    y,
    marker='o',
    linewidth=2
)

ax.set_title('Contoh Grafik')

ax.set_xlabel('Sumbu X')

ax.set_ylabel('Sumbu Y')

ax.grid(True)

plt.show()
```

```python
kategori = [
    'Makanan',
    'Fashion',
    'Elektronik',
    'Kesehatan',
    'Hiburan'
]

val_2024 = [42,31,28,19,14]

val_2025 = [48,27,35,25,18]

x = np.arange(len(kategori))

width = 0.35

fig, ax = plt.subplots(figsize=(10,6))

bars1 = ax.barh(
    x - width/2,
    val_2024,
    width,
    label='2024'
)

bars2 = ax.barh(
    x + width/2,
    val_2025,
    width,
    label='2025'
)

ax.set_yticks(x)

ax.set_yticklabels(kategori)

ax.set_xlabel('Pengeluaran')

ax.set_title('Perbandingan Pengeluaran')

ax.legend()

ax.bar_label(bars1)

ax.bar_label(bars2)

plt.show()
```

What?
Grafik menunjukkan perbandingan pengeluaran beberapa kategori antara tahun 2024 dan 2025.

So what?
Kategori makanan memiliki pengeluaran tertinggi pada kedua tahun.
Terjadi kenaikan pada kategori elektronik dan kesehatan di tahun 2025.

Now what?
Perusahaan dapat meningkatkan stok atau promosi pada kategori dengan pertumbuhan tertinggi seperti elektronik.

```python
bulan = [
    'Jan','Feb','Mar','Apr','Mei','Jun',
    'Jul','Agu','Sep','Okt','Nov','Des'
]

kota_a = [
    22,25,28,32,38,42,
    45,43,39,34,28,24
]

kota_b = [
    15,17,22,26,30,35,
    38,37,32,27,20,16
]

fig, ax = plt.subplots(figsize=(10,6))

ax.plot(
    bulan,
    kota_a,
    marker='o',
    linewidth=2,
    label='Kota A'
)

ax.plot(
    bulan,
    kota_b,
    marker='s',
    linewidth=2,
    linestyle='--',
    label='Kota B'
)

ax.fill_between(
    bulan,
    kota_a,
    kota_b,
    alpha=0.2
)

ax.set_title('Suhu Rata-rata')

ax.set_ylabel('Suhu')

ax.legend()

ax.grid(True)

plt.show()
```

What?
Grafik menunjukkan perubahan suhu rata-rata bulanan untuk Kota A dan Kota B selama satu tahun.

So what?
Kota A memiliki suhu lebih tinggi dibanding Kota B hampir di seluruh bulan.
Puncak suhu terjadi pada pertengahan tahun.

Now what?
Data ini dapat digunakan untuk analisis iklim musiman dan perencanaan aktivitas berdasarkan cuaca.

```python
fig, ax = plt.subplots(figsize=(9,6))

sns.scatterplot(
    data=df,
    x='sepal_length',
    y='petal_length',
    hue='species',
    size='petal_width',
    sizes=(30,200),
    ax=ax
)

sns.regplot(
    data=df,
    x='sepal_length',
    y='petal_length',
    scatter=False,
    color='gray',
    ax=ax
)

ax.set_title('Scatter Plot Iris')

plt.show()
```

What?
Grafik menunjukkan hubungan antara sepal_length dan petal_length pada dataset iris.

So what?
Terdapat hubungan positif antara kedua variabel.
Semakin besar sepal_length, cenderung semakin besar petal_length.

Now what?
Variabel ini dapat digunakan sebagai fitur penting untuk klasifikasi spesies bunga iris.

```python
fig, ax = plt.subplots(figsize=(8,5))

sns.histplot(
    df['sepal_length'],
    bins=20,
    kde=True,
    ax=ax
)

ax.set_title('Histogram Sepal Length')

plt.show()
```

What?
Histogram menunjukkan distribusi nilai sepal_length pada dataset iris.

So what?
Sebagian besar data berada pada rentang tengah dan distribusi terlihat mendekati normal.

Now what?
Distribusi ini menunjukkan bahwa data cukup stabil untuk digunakan pada analisis statistik lanjutan.

```python
tips = sns.load_dataset('tips')

fig, ax = plt.subplots(figsize=(9,5))

sns.histplot(
    data=tips,
    x='total_bill',
    hue='time',
    kde=True,
    palette='Set2',
    alpha=0.6,
    ax=ax
)

ax.set_title('Distribusi Total Bill')

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8,5))

sns.kdeplot(
    data=tips,
    x='total_bill',
    hue='sex',
    fill=True,
    alpha=0.4,
    ax=ax
)

ax.set_title('KDE Total Bill')

plt.show()
```

What?
Grafik KDE menunjukkan distribusi total_bill berdasarkan jenis kelamin pelanggan.

So what?
Distribusi pelanggan laki-laki memiliki rentang nilai yang lebih besar dibanding perempuan.

Now what?
Restoran dapat melakukan segmentasi pelanggan berdasarkan pola pengeluaran.

```python
fig, ax = plt.subplots(figsize=(8,5))

sns.boxplot(
    data=df,
    x='species',
    y='petal_length',
    palette='Set2',
    ax=ax
)

ax.set_title('Boxplot Petal Length')

plt.show()
```

What?
Boxplot memperlihatkan distribusi petal_length berdasarkan spesies iris.

So what?
Species setosa memiliki petal paling pendek dan konsisten.
Virginica memiliki petal paling panjang.

Now what?
Petal_length dapat digunakan sebagai indikator penting dalam identifikasi spesies iris.

```python
fig, ax = plt.subplots(figsize=(8,5))

sns.violinplot(
    data=df,
    x='species',
    y='petal_length',
    palette='Set2',
    inner='box',
    ax=ax
)

ax.set_title('Violin Plot')

plt.show()
```

What?
Violin plot menunjukkan distribusi lengkap petal_length untuk setiap spesies iris.

So what?
Distribusi virginica lebih lebar dibanding setosa sehingga variasi datanya lebih besar.

Now what?
Analisis distribusi membantu memahami karakteristik tiap spesies secara lebih detail.

```python
fig, ax = plt.subplots(figsize=(8,5))

sns.stripplot(
    data=df,
    x='species',
    y='petal_length',
    jitter=True,
    alpha=0.6,
    palette='Set2',
    ax=ax
)

ax.set_title('Strip Plot')

plt.show()
```

What?
Strip plot menampilkan seluruh titik data petal_length dari masing-masing spesies iris.

So what?
Data setosa terkumpul rapat sedangkan virginica lebih menyebar.

Now what?
Perbedaan penyebaran data dapat digunakan untuk analisis variasi antar spesies.

```python
pair = sns.pairplot(
    df,
    hue='species',
    palette='Set2',
    diag_kind='kde',
    plot_kws={'alpha':0.6}
)

pair.fig.suptitle(
    'Pair Plot Iris',
    y=1.02
)

plt.show()
```

What?
Pair plot menunjukkan hubungan seluruh variabel numerik pada dataset iris.

So what?
Petal_length dan petal_width memiliki korelasi paling kuat.
Species setosa terlihat paling mudah dipisahkan dibanding spesies lain.

Now what?
Variabel petal sangat potensial digunakan untuk model machine learning klasifikasi iris.
