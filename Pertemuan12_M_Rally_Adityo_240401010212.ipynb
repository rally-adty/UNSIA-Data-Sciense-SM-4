Install library

```python
import pandas as pd
import matplotlib.pyplot as plt

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

Import library

```python
transaksi = [
    ['Roti','Susu','Telur'],
    ['Roti','Selai'],
    ['Susu','Telur','Keju'],
    ['Roti','Susu','Selai'],
    ['Roti','Telur'],
    ['Susu','Sereal']
]
```

Membuat data transaksi

```python
transaksi
```

Melihat data

```python
te = TransactionEncoder()
```

```python
te_array = te.fit(transaksi).transform(transaksi)
```

```python
df = pd.DataFrame(
    te_array,
    columns=te.columns_
)
```

One hot encoding

```python
df.head()
```

Menampilkan dataset

```python
freq_items = apriori(
    df,
    min_support=0.3,
    use_colnames=True
)
```

Frequent itemset

```python
freq_items = freq_items.sort_values(
    by="support",
    ascending=False
)
```

Mengurutkan frequent itemset

```python
freq_items
```

Menampilkan frequent itemset

```python
rules = association_rules(
    freq_items,
    metric="confidence",
    min_threshold=0.5
)
```

Membuat associaton rules

```python
rules = rules[
    rules["lift"] > 1
]
```

Filter lift

```python
rules = rules.sort_values(
    by="lift",
    ascending=False
)
```

Mengurutkan lift

```python
rules[
    [
        "antecedents",
        "consequents",
        "support",
        "confidence",
        "lift"
    ]
]
```

Menampilkan rules

```python
top = freq_items.sort_values(
    by="support"
).tail(5)
```

menampilkan top 5 itemset

```python
plt.figure(figsize=(8,4))
```

```python
plt.barh(
    top["itemsets"].astype(str),
    top["support"]
)
```

```python
plt.xlabel("Support")
```

```python
plt.title("Top 5 Frequent Itemset")
```

```python
plt.show()
```
